## 8. Filter rows based on a condition

```python
# Students who scored more than 70 in Math
high_math = students[students["Math"] > 70]
print("Students with Math > 70:", len(high_math))
high_math[["Student_ID", "Name", "Math"]].head()
```

```python
# Multiple conditions: Math > 60 AND Science > 60 (use & and brackets)
both = students[(students["Math"] > 60) & (students["Science"] > 60)]
print("Students with Math > 60 and Science > 60:", len(both))
both[["Name", "Math", "Science"]].head()
```

```python
# Filter using a text condition
females = students[students["Gender"] == "Female"]
print("Number of female students:", len(females))
```
