# Week 2 - Task 1: NumPy Fundamentals
**What is NumPy?** NumPy (Numerical Python) is the core Python library for numerical computing. Its main object is the `ndarray`, a fast, fixed-type, n-dimensional array. NumPy arrays use less memory and are much faster than Python lists because operations are vectorised (applied to the whole array at once without writing loops). Almost every AI/ML library (Pandas, Scikit-learn, TensorFlow, PyTorch) is built on top of NumPy, so it is the foundation of data handling in AI.

```python
import numpy as np
print("NumPy version:", np.__version__)
```

## 1. Creating NumPy arrays (1D and 2D)

```python
# 1D array (vector) - marks of 5 students
marks = np.array([45, 78, 62, 90, 33])
print("1D array:", marks)

# 2D array (matrix) - 3 students x 3 subjects
scores = np.array([[70, 65, 80],
                   [55, 60, 58],
                   [90, 85, 95]])
print("2D array:\n", scores)

# Other ways of creating arrays
print("zeros  :", np.zeros(4))
print("ones   :\n", np.ones((2, 3)))
print("arange :", np.arange(0, 10, 2))
print("linspace:", np.linspace(0, 1, 5))
```

## 2. Shape and dimensions

```python
print("marks  -> ndim:", marks.ndim, "| shape:", marks.shape, "| size:", marks.size, "| dtype:", marks.dtype)
print("scores -> ndim:", scores.ndim, "| shape:", scores.shape, "| size:", scores.size, "| dtype:", scores.dtype)

# reshape changes the shape without changing the data
print("Reshaped (1x5 -> 5x1):\n", marks.reshape(5, 1))
```

## 3. Indexing and slicing

```python
print("First element   :", marks[0])
print("Last element    :", marks[-1])
print("Slice [1:4]     :", marks[1:4])
print("Every 2nd mark  :", marks[::2])

print("\nscores[0, 1]     :", scores[0, 1], "(row 0, column 1)")
print("First row       :", scores[0])
print("Second column   :", scores[:, 1])
print("Top-left 2x2    :\n", scores[:2, :2])

# Boolean indexing - select marks above 60
print("\nMarks above 60  :", marks[marks > 60])
```

## 4. Basic mathematical operations

```python
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])
print("a + b  :", a + b)
print("a - b  :", a - b)
print("a * b  :", a * b)
print("a / b  :", a / b)
print("a ** 2 :", a ** 2)
print("a + 5  :", a + 5, "(broadcasting - scalar applied to every element)")

# Matrix multiplication
m1 = np.array([[1, 2], [3, 4]]); m2 = np.array([[5, 6], [7, 8]])
print("\nMatrix product:\n", m1 @ m2)
```

## 5. mean(), max(), min() and sum()

```python
print("Marks        :", marks)
print("Mean         :", marks.mean())
print("Maximum      :", marks.max())
print("Minimum      :", marks.min())
print("Sum          :", marks.sum())

print("\nscores (rows = students, columns = subjects):\n", scores)
print("Mean of each subject (axis=0) :", scores.mean(axis=0))
print("Total of each student (axis=1):", scores.sum(axis=1))
print("Highest overall mark          :", scores.max())
print("Lowest overall mark           :", scores.min())
```

## 6. Small AI-related example: finding the best student

```python
names = np.array(["Asha", "Ravi", "Meena"])
avg = scores.mean(axis=1)
print("Average per student:", avg.round(2))
print("Topper:", names[avg.argmax()], "with", avg.max().round(2))
print("Students who passed (avg >= 60):", names[avg >= 60])
```
